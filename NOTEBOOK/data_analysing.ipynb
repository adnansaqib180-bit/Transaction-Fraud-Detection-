```python
normal_df = df[df['is_fraud'] == 0] 
fraud_df = df[df['is_fraud'] == 1]  
```

```python
final_df = pd.concat([normal_df_samples, fraud_df], axis=0)
final_df = final_df.sample(frac=1, random_state=42).reset_index(drop=True)
```
