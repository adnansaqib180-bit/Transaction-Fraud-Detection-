```python
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd 
```

```python
df =  pd.read_csv('../DATA/processed_balanced_data.csv')
```

```python
print(df.columns)
```

```python
X   =   df.drop('is_fraud', axis=1)
Y  =  df['is_fraud']
import numpy as np 
X['amt'] = np.log1p(X['amt'])
```

```python
sns.countplot(x='is_fraud', data=df,hue='gender')
```

```python
plt.figure(figsize=(6,6))
sns.boxplot(x=Y, y='amt', data=X)
plt.show()
```

```python
sns.countplot(x='is_fraud', data=df)
```

```python
plt.figure(figsize=(10,6))
sns.countplot(y=X['category'],hue=Y[Y == 1])
plt.tight_layout()
```

```python
sns.kdeplot(data=df, x='city_pop', hue='is_fraud', fill=True, common_norm=False, palette='Set1', alpha=0.5)
```
